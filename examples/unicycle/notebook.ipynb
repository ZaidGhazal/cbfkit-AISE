```python
import pandas as pd

df= pd.read_csv('/home/cbfkit/examples/unicycle/start_to_goal/results/operational_rule_datasets_200_samples_either/unicycle_dynamic_obstacle/D_paired_comparison.csv')
```

```python
## Count where evolved_label is Pass and legacy_label is Fail
ev_pass_leg_fail = df[(df["evolved_label"] == "Pass") & (df["legacy_label"] == "Fail")].shape[0]
ev_fail_leg_pass = df[(df["evolved_label"] == "Fail") & (df["legacy_label"] == "Pass")].shape[0]
ev_pass_leg_pass = df[(df["evolved_label"] == "Pass") & (df["legacy_label"] == "Pass")].shape[0]
ev_fail_leg_fail = df[(df["evolved_label"] == "Fail") & (df["legacy_label"] == "Fail")].shape[0]
print(f"Evolved Pass & Legacy Fail: {ev_pass_leg_fail}")
print(f"Evolved Fail & Legacy Pass: {ev_fail_leg_pass}")
print(f"Evolved Pass & Legacy Pass: {ev_pass_leg_pass}")
print(f"Evolved Fail & Legacy Fail: {ev_fail_leg_fail}")

## Calculate the percentage of cases
total_cases = df.shape[0]
ev_pass_leg_fail_percentage = (ev_pass_leg_fail / total_cases) * 100
ev_fail_leg_pass_percentage = (ev_fail_leg_pass / total_cases) * 100
ev_pass_leg_pass_percentage = (ev_pass_leg_pass / total_cases) * 100
ev_fail_leg_fail_percentage = (ev_fail_leg_fail / total_cases) * 100
print(f"Percentage of cases where Evolved Pass & Legacy Fail: {ev_pass_leg_fail_percentage:.2f}%")
print(f"Percentage of cases where Evolved Fail & Legacy Pass: {ev_fail_leg_pass_percentage:.2f}%")
print(f"Percentage of cases where Evolved Pass & Legacy Pass: {ev_pass_leg_pass_percentage:.2f}%")
print(f"Percentage of cases where Evolved Fail & Legacy Fail: {ev_fail_leg_fail_percentage:.2f}%")


## Number and percentage of total cases where Evolved Passes
total_ev_pass = ev_pass_leg_fail + ev_pass_leg_pass
total_ev_fail = ev_fail_leg_pass + ev_fail_leg_fail
total_ev_pass_percentage = (total_ev_pass / total_cases) * 100
total_ev_fail_percentage = (total_ev_fail / total_cases) * 100
print(f"Total cases where Evolved Passes: {total_ev_pass} ({total_ev_pass_percentage:.2f}%)")
print(f"Total cases where Evolved Fails: {total_ev_fail} ({total_ev_fail_percentage:.2f}%)")

## Number and percentage of total cases where Legacy Passes
total_leg_pass = ev_pass_leg_pass + ev_fail_leg_pass
total_leg_fail = ev_pass_leg_fail + ev_fail_leg_fail
total_leg_pass_percentage = (total_leg_pass / total_cases) * 100
total_leg_fail_percentage = (total_leg_fail / total_cases) * 100
print(f"Total cases where Legacy Passes: {total_leg_pass} ({total_leg_pass_percentage:.2f}%)")
print(f"Total cases where Legacy Fails: {total_leg_fail} ({total_leg_fail_percentage:.2f}%)")
```
